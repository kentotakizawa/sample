```python
import pandas as pd
import re
from typing import List
import numpy as np
```

```python
# 重要概念数1のデータを除外した場合
```

```python
df = pd.read_excel('/home/takizawa/2025_takizawa/2026_research/webdb_exp/x_Results/_0_2_実験2.xlsx', sheet_name='result1_export_add')
# df.head()
```

```python
# 除外するインデックスのリストを作成する
remove_index = [2, 17, 22, 28, 29, 31, 33, 52, 60, 73, 76, 81, 90]
```

```python
# dfに新しい列「エッジリスト」「推定された理解度」「推定された理解不足の概念」を追加する
df['エッジリスト'] = None
df['推定された理解度'] = None
df['推定された理解不足の概念'] = None
```

```python
# 新しい列に対して値を追加する
#.get()メソッドは，辞書から概要するキーの値を取得し，存在しない場合はNoneを返す

for index in range(len(df)):
    if index in remove_index:
        continue

    else:
        selected_FileName = f"/home/takizawa/2025_takizawa/2026_research/webdb_exp/x_Results/exp_results_{index}.pkl"
        selected_dic = pd.read_pickle(selected_FileName)

        df.at[index, 'エッジリスト'] = selected_dic.get('graph_edges', None)
        df.at[index, '推定された理解度'] = selected_dic.get('understanding', None)
        df.at[index, '推定された理解不足の概念'] = selected_dic.get('min_understanding_node', None)


# df.head()
```

```python
# 1. 理解不足の概念の予測との一致度を計算するために，列「理解不足単語集合」をリストに変換する

def clean_and_split_string(input_data):
    """
    入力された文字列から，指定された区切り文字を全て除去し，リストに分割する。
    """
    # --- 1. 文字列への強制変換 ---
    # ここで str() を使うことで、floatやintなどの型をstrに変換し、エラーを回避する
    input_str = str(input_data)

    # --- 2. 正規表現処理 ---
    cleaned_str = re.sub(r'[,\s|]+', ' ', input_str).strip()

    # --- 3. リストの作成 --- 
    result_list = cleaned_str.split(' ')

    return result_list

# applyメソッドを使って、各行の文字列を処理する
df['理解不足単語集合'] = df['理解不足単語集合'].apply(clean_and_split_string)


# df.head()
```

```python
# 2. 理解不足の概念の一致を計算するために，リスト型を文字列型に加工する
def convert_to_string(item):
    """
    要素の型をチェックし、リスト型であれば文字列に変換する処理
    """
    # 要素がリスト型の場合
    if isinstance(item, list):
        # 解決策1: リスト全体を文字列の表現にする
        return str(item[0]) 
        
        # ▼ 【別解】もし、リストの要素を「, 」で繋げたい場合は、こちらを使用してください ▼
        # return ", ".join(map(str, item))
    
    # それ以外の型 (str, float, Noneなど) の場合はそのまま返す
    return item


# .apply() メソッドを使って、列のすべての要素にカスタム関数を適用する
df['推定された理解不足の概念'] = df['推定された理解不足の概念'].apply(convert_to_string)
```

```python
# 3. 理解不足の概念の一致を計算し，新しい列「理解不足の概念の一致」に追加する
# df.appyly()は，DataFrame全体に特定の関数を適用するためのメソッド

def check_if_in_list(row):
    if row['推定された理解不足の概念'] in row['理解不足単語集合']:
        return 1
    else:
        return 0

df['理解不足の概念の一致'] = df.apply(check_if_in_list, axis=1) # axis=1 を指定することで，関数が行全体 (Series形式)を受け取る


# df.head()
```

```python
# ファイルを保存する
df.to_excel('/home/takizawa/2025_takizawa/2026_research/webdb_exp/x_Results/results_0_removed.xlsx', index=False) 
```

```python
# 精度を計算する

# 人手による理解度と推定されたそれの順位相関係数
understanding_corr_simple = df['推定された理解度'].corr(df['simple_average'], method='spearman')
print(f"順位相関係数(比較:simple_average): {understanding_corr_simple}")

understanding_corr_relativize = df['推定された理解度'].corr(df['相対化_average'], method='spearman')
print(f"順位相関係数(比較:相対化_average): {understanding_corr_relativize}")


# 人手による理解不足の概念と推定のそれの一致度
lack_understanding_mean = df['理解不足の概念の一致'].mean()
print(f"一致度: {lack_understanding_mean}")
```
